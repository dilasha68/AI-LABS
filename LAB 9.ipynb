# Convolutional Neural Network (CNN) — Theory

A **Convolutional Neural Network (CNN)** is a type of deep neural network that is especially effective for processing data with a spatial structure, such as images, video frames, and other grid-based data. CNNs automatically identify useful features from the input by applying an operation known as **convolution**.

Unlike fully connected neural networks, where every neuron is connected to every neuron in the next layer, CNNs focus on small local regions of the input and reuse the same parameters across different locations. This reduces the number of parameters and allows CNNs to efficiently learn visual features such as edges, textures, shapes, and objects.

---

## 1. Convolution

**Convolution** is the main operation used by a CNN to extract useful information from an input. It involves moving a small matrix called a **kernel** or **filter** across the input image.

At each position, the values of the kernel are multiplied element-by-element with the corresponding values of the image. These products are then added together to generate one value in the resulting **feature map**.

### Why is Convolution Used?

Convolution is useful because it:

* Identifies local features such as edges, corners, textures, and shapes.
* Uses the same filter at different positions in the image, known as **parameter sharing**.
* Requires significantly fewer parameters than a fully connected neural network.
* Maintains important spatial relationships between neighboring pixels.

---

## 2. Main Layers of a CNN

A CNN generally consists of several different layers that perform specific operations. The major layers include:

1. **Convolutional Layer** — detects and extracts important features from the input.
2. **Pooling Layer** — reduces the spatial dimensions of feature maps.
3. **Batch Normalization Layer** — normalizes activations to make training more stable.
4. **Dropout Layer** — reduces overfitting by randomly disabling some neurons during training.

CNN architectures also commonly contain **activation functions** such as ReLU and **fully connected layers** near the output for classification.

---

## 3. Convolutional Layer

The **convolutional layer** is responsible for extracting features from an image using a collection of learnable filters.

An image can be represented as a three-dimensional tensor containing its **channels, height, and width**.

For example:

* Grayscale image → `(1, H, W)`
* RGB image → `(3, H, W)`
* CMYK image → `(4, H, W)`

Image pixels commonly contain values between 0 and 255. Before training, these values can be normalized to a smaller range such as `[0,1]` to make the training process more stable.

### Kernel or Filter

A **kernel** is a small matrix that moves across an image to detect particular features. Common kernel sizes include **3 × 3** and **5 × 5**.

The numbers contained inside a kernel are not normally fixed manually. Instead, they are learned during training through **backpropagation**.

Different filters can learn to respond to different characteristics of an image. For example, one filter may become sensitive to vertical edges while another may identify horizontal edges or textures.

The convolution operation can be represented as:

$$
\text{output}(i,j)
==================

\sum_m\sum_n
\text{image}(i+m,j+n)
\cdot
\text{kernel}(m,n)
$$

The resulting matrix is called a **feature map**.

### Weights

The numerical values contained within each kernel are called **weights**. These weights are trainable parameters.

During training, the network repeatedly modifies these values in order to reduce the error calculated by the loss function. As a result, filters gradually learn which visual patterns are useful for solving the given task.

---

## 4. Padding

**Padding** means adding additional pixels around the boundary of an input before applying convolution. In most cases, these additional values are zeros, which is known as **zero padding**.

Without padding, the spatial dimensions of a feature map become smaller whenever convolution is performed. Pixels near the boundary of an image also participate in fewer convolution operations than pixels near the center.

### Why Use Padding?

Padding can:

* Prevent feature maps from shrinking too quickly.
* Preserve information located near the edges of an image.
* Allow control over the spatial dimensions of the output.

Two commonly used padding methods are:

### Valid Padding

**Valid padding** means that no additional pixels are placed around the input.

Therefore, the output feature map is usually smaller than the original input.

### Same Padding

**Same padding** adds enough pixels around the input so that the output can retain the same height and width as the input when the stride is 1.

---

## 5. Stride

**Stride** represents the number of pixels by which a filter moves each time it slides across an image.

For example:

* **Stride = 1:** The filter moves one pixel at a time.
* **Stride = 2:** The filter moves two pixels at a time.

A smaller stride examines the image more closely and produces a larger feature map. A larger stride reduces the size of the feature map and decreases computational requirements, but some spatial information may be lost.

### Output Size Formula

For an input dimension $W$, kernel size $K$, padding $P$, and stride $S$, the output dimension is calculated as:

$$
\text{Output Size}
==================

\left\lfloor
\frac{W-K+2P}{S}
\right\rfloor + 1
$$

The floor operation is used because the output dimension must be an integer.

For an odd-sized kernel and stride of 1, padding that preserves the input dimension is commonly:

$$
P = \frac{K-1}{2}
$$

For example, a **3 × 3 kernel** normally requires padding of **1 pixel** on each side to preserve the spatial dimensions.

---

## 6. Pooling Layer

A **pooling layer** reduces the height and width of feature maps while attempting to retain their most useful information.

A small window, commonly **2 × 2**, moves across the feature map and combines the values within each region.

Pooling helps to:

* Reduce the dimensions of feature maps.
* Lower computational requirements in later layers.
* Reduce sensitivity to small changes in feature position.
* Help reduce overfitting in some architectures.

Two common pooling techniques are **max pooling** and **average pooling**.

### Max Pooling

**Max pooling** selects the largest value from each pooling window.

For example:

$$
\begin{bmatrix}
2 & 7\
4 & 3
\end{bmatrix}
\rightarrow 7
$$

This allows the strongest activation within that region to be preserved.

### Average Pooling

**Average pooling** calculates the average of all values within the pooling region.

For example:

$$
\begin{bmatrix}
2 & 6\
4 & 8
\end{bmatrix}
\rightarrow
\frac{2+6+4+8}{4}=5
$$

Average pooling therefore provides a more generalized summary of the region.

---

## 7. Batch Normalization

**Batch Normalization (BatchNorm)** is a technique used to normalize intermediate activations during neural network training.

It can make optimization more stable and often allows a network to train faster.

For a mini-batch containing $m$ values, the mean is calculated as:

$$
\mu =
\frac{1}{m}
\sum_{i=1}^{m}x_i
$$

The variance is:

$$
\sigma^2 =
\frac{1}{m}
\sum_{i=1}^{m}(x_i-\mu)^2
$$

The values are then normalized:

$$
\hat{x}_i =
\frac{x_i-\mu}
{\sqrt{\sigma^2+\epsilon}}
$$

Here, $\epsilon$ is a small constant used to prevent division by zero.

After normalization, two learnable parameters are applied:

* $\gamma$ — controls scaling.
* $\beta$ — controls shifting.

The final output becomes:

$$
y_i = \gamma\hat{x}_i+\beta
$$

Because $\gamma$ and $\beta$ are learned during training, the network can determine the most appropriate scale and offset for the normalized activations.

### Benefits of Batch Normalization

Batch normalization can:

* Improve training stability.
* Allow the use of larger learning rates in many cases.
* Speed up convergence.
* Provide a small regularization effect in some situations.

---

## 8. Dropout

**Dropout** is a regularization method used to reduce overfitting.

During training, a selected percentage of activations are randomly set to zero. As a result, the model cannot depend excessively on particular neurons and is encouraged to learn more distributed representations.

For example, if:

$$
\text{Dropout Rate}=0.5
$$

approximately 50% of the selected activations are dropped during each training step.

The neurons that are removed vary randomly between training steps.

Dropout behaves differently during inference. When the trained model is used to make predictions, dropout is disabled and the complete network is used, with the framework accounting for the appropriate activation scaling.

---

## 9. Activation Function

After convolution, CNNs generally apply a nonlinear **activation function**.

One of the most widely used activation functions in CNNs is **ReLU (Rectified Linear Unit)**.

It is defined as:

$$
f(x)=\max(0,x)
$$

Therefore:

* If $x > 0$, the output is $x$.
* If $x \leq 0$, the output is $0$.

ReLU introduces non-linearity into the network, allowing it to learn complex relationships that could not be represented by a sequence of purely linear operations.

---

## 10. Fully Connected Layer

After several convolutional and downsampling stages, the extracted features are eventually used to produce the final prediction.

In traditional CNN architectures, feature maps may first be **flattened** into a one-dimensional vector and then passed through one or more **fully connected (dense) layers**.

For example:

$$
(64,7,7)
\rightarrow
64\times7\times7
================

3136
$$

The resulting 3136 values can then be supplied to a dense layer.

Modern CNN architectures may instead use techniques such as **global average pooling**, which can significantly reduce the number of parameters before the final output layer.

---

## 11. Output Layer

The final layer depends on the type of problem being solved.

For **multi-class classification**, the **Softmax** function is commonly used.

Softmax converts output scores into probabilities:

$$
P(y=i)=
\frac{e^{z_i}}
{\sum_j e^{z_j}}
$$

The predicted class is generally the class with the highest probability.

For **binary classification**, a single output neuron with a **sigmoid** activation is commonly used instead.

---

## 12. Typical CNN Architecture

A basic CNN for image classification may follow this structure:

**Input Image**

↓

**Convolution Layer**

↓

**Batch Normalization**

↓

**ReLU Activation**

↓

**Max Pooling**

↓

**Convolution Layer**

↓

**Batch Normalization**

↓

**ReLU Activation**

↓

**Max Pooling**

↓

**Flatten or Global Average Pooling**

↓

**Fully Connected Layer**

↓

**Dropout**

↓

**Output Layer**

The early convolutional layers generally detect basic patterns such as edges and lines. Intermediate layers combine these patterns into textures and shapes, while deeper layers can learn more abstract and task-specific features.

---

## 13. How a CNN Learns

CNNs learn through repeated training using **forward propagation and backpropagation**.

The general training process is:

1. Input data is passed through the CNN.
2. Convolutional layers generate feature maps.
3. Activation and pooling operations process the extracted features.
4. The network produces a prediction.
5. A **loss function** measures the difference between the prediction and the actual target.
6. **Backpropagation** calculates how each trainable parameter contributed to the error.
7. An optimizer such as **SGD** or **Adam** updates the weights.
8. The process is repeated over many batches and epochs.

As training continues, the filters become increasingly useful for recognizing patterns related to the task.

---

## 14. Advantages of CNNs

CNNs provide several advantages when working with image and spatial data:

* **Automatic feature extraction:** Features do not need to be manually designed.
* **Parameter sharing:** The same kernel is applied across multiple spatial locations.
* **Local connectivity:** Filters focus on nearby regions of the input.
* **Hierarchical learning:** Simple features can gradually be combined into complex representations.
* **Efficient image processing:** CNNs generally require fewer parameters than similarly sized fully connected networks operating directly on images.
* **Wide applicability:** CNNs can be used for classification, detection, segmentation, medical imaging, video analysis, and many other computer vision problems.

---

## 15. Conclusion

A **Convolutional Neural Network** is a deep learning model that learns spatial features by repeatedly applying convolutional filters to input data. Kernels extract local patterns, while padding and stride control how convolution moves across the input and determine the dimensions of the resulting feature maps.

Pooling can reduce spatial dimensions, Batch Normalization can improve training stability, and Dropout can help prevent overfitting. Activation functions such as ReLU allow the network to model nonlinear relationships, while dense layers or global pooling can convert the learned features into final predictions.

By combining these operations, CNNs can progressively learn representations ranging from simple edges and textures to complex shapes and objects, making them one of the foundational architectures used in computer vision.
