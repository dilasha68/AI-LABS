## Theory

### 1. Semantic Network

A semantic network represents the knowledge in terms of graph data structures. In the network, objects, concepts, or entities can be represented by the nodes, while their relationships with each other can be represented by the edges.

For instance, in a semantic network, "Apple" may have a link with "Fruit" using "is a" relation. Using such a relation will enable computers to understand that "apple" is a type of fruit.

### 2. Part-of-Speech Tagging

Part-of-Speech tagging is a natural language processing technique that allows identifying the function of each word in a sentence in terms of their grammatical role.

For example, in the sentence "Ram eats an apple," "Ram" is a noun, "eats" is a verb, and "apple" is again a noun. Such tagging is necessary for understanding the sentence structure.

### 3. Named Entity Recognition

Named Entity Recognition, or NER for short, is a technique that allows to recognize real-world entities from text data. Such entities may include names of persons, companies, locations, dates, and other relevant entities.

For instance, if we consider the statement “Google is located in California”, NER would help in identifying "Google" as an organization and "California" as a location.

### 4. Vector Embedding

Vector embedding is the process of transforming words into numerical format. Because computers cannot perceive words directly, NLP models encode words by using vectors – that is, strings of numbers.

Words that share meaning often share similarity in vector space too. Words like “king” and “queen” are likely to be close to one another in vector space as they share common meaning.

### 5. Knowledge Triples

Knowledge triples serve to represent information in three elements - subject, predicate, and object.

**Example:**

Apple -> is a -> Fruit

In this example, "Apple" is the subject, "is a" is the predicate, and "Fruit" is the object.

### 6. Recommendation System

A recommendation system suggests items to users based on similarity or past behavior.

Semantic networks may assist recommendation systems in identifying the relations among users, products, and preferences.

For instance, if two users have similar purchases, the recommendation system may suggest the other user items that appear in the first user's shopping basket.

### 7. Jaccard Similarity

Jaccard Similarity is applied to calculate the degree of similarity between two sets. It equals the number of common elements divided by the sum of all distinct elements.

**Formula:**

Jaccard Similarity = Intersection/Union

**Example:**

Cart A: {Banana, Apple, Bread}  
Cart B: {Banana, Apple, Orange}

Number of common items: 2  
Total number of items: 4

Then,

Jaccard Similarity = 2/4 = 0.5 = 50%

It means that there are 50% similarities between these two carts.

### 8. Cosine Similarity

Cosine similarity allows measuring the similarities between vectors by checking whether they point in the same direction.

Cosine similarity may be helpful in natural language processing because it allows determining the similarities between texts or vectors created on their basis.
