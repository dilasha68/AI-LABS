## Theory

### 1. Machine Learning

Machine Learning is a branch of Artificial Intelligence that allows computers to learn from data and improve their performance without being explicitly programmed for every task. In traditional programming, humans write fixed rules and instructions for the computer. However, in Machine Learning, the computer is trained using data so that it can identify patterns and make predictions or decisions. For example, a machine learning model can learn from previous email data and predict whether a new email is spam or not spam. Machine Learning is widely used in recommendation systems, medical diagnosis, image recognition, speech recognition, fraud detection, robotics, and self-driving vehicles.

### 2. Types of Machine Learning

Machine Learning can be divided into different types based on how the model learns from data. The main types are supervised learning, unsupervised learning, semi-supervised learning, self-supervised learning, and reinforcement learning. Each type uses a different learning method depending on the type of data available and the problem that needs to be solved. Some methods require labelled data, some work with unlabelled data, and some learn through rewards and penalties.

### 3. Supervised Learning

Supervised learning is a type of Machine Learning where the model is trained using labelled data. Labelled data means the input data already contains the correct output. The model learns the relationship between input and output and then uses this knowledge to predict the output for new data. Supervised learning is commonly used for classification and regression problems. Examples include spam email detection, disease prediction, student grade prediction, and house price prediction.

### 4. Unsupervised Learning

Unsupervised learning is a type of Machine Learning where the model is trained using unlabelled data. In this method, the data does not contain correct answers or labels. The model tries to find hidden patterns, groups, or structures in the data by itself. Unsupervised learning is mainly used for clustering and pattern discovery. Examples include customer segmentation, grouping similar documents, anomaly detection, and market basket analysis.

### 5. Semi-Supervised Learning

Semi-supervised learning is a type of Machine Learning that uses both labelled and unlabelled data. Usually, only a small amount of data is labelled, while a large amount of data is unlabelled. This method is useful when labelling data is expensive, difficult, or time-consuming. For example, in medical image classification, only a few images may be labelled by doctors, while many other images remain unlabelled. Semi-supervised learning helps the model learn better by using both labelled and unlabelled data.

### 6. Self-Supervised Learning

Self-supervised learning is a type of Machine Learning where the model creates its own labels from the available data. It does not require manually labelled data. The model learns by solving tasks created from the data itself. For example, in natural language processing, a word can be removed from a sentence and the model learns to predict the missing word. Self-supervised learning is widely used in modern Artificial Intelligence systems, especially in language models, computer vision, and speech recognition.

### 7. Reinforcement Learning

Reinforcement learning is a type of Machine Learning where an agent learns by interacting with an environment. The agent performs actions and receives rewards or penalties based on the result of those actions. The main goal is to learn the best actions that give the maximum reward over time. Reinforcement learning is different from other types because it is based on trial and error. It is commonly used in robotics, game playing, self-driving vehicles, and automated decision-making systems.

### 8. Machine Learning Life Cycle

The Machine Learning life cycle is the step-by-step process used to build a machine learning model. It includes data collection, data preprocessing, model training, model testing, and model evaluation. First, useful data is collected from different sources. Then, the data is cleaned and prepared for use. After preprocessing, the model is trained using the dataset. The trained model is then tested with new data, and its performance is evaluated using different evaluation metrics.

### 9. Data Collection

Data collection is the process of gathering data required for building a machine learning model. The quality and quantity of data directly affect the accuracy and performance of the model. Data can be collected from different sources such as databases, files, surveys, sensors, websites, or user records. A good dataset should be accurate, relevant, complete, and suitable for the problem being solved. If the collected data is poor, the model may produce incorrect predictions.

### 10. Data Preprocessing

Data preprocessing is the process of cleaning and preparing raw data before using it in a machine learning model. Raw data may contain missing values, duplicate records, incorrect values, or unnecessary information. Data preprocessing includes data cleaning, handling missing values, removing unwanted data, converting data into the correct format, and normalization. Normalization is used to bring data values into a similar range so that the model can learn more effectively.

### 11. Classification and Regression

Classification and regression are two common types of supervised learning problems. Classification is used when the output is a category or class. For example, predicting whether an email is spam or not spam is a classification problem. Regression is used when the output is a continuous numerical value. For example, predicting house price, temperature, or sales amount is a regression problem. Gaussian Naive Bayes is mainly used for classification problems.

### 12. Gaussian Naive Bayes

Gaussian Naive Bayes is a supervised Machine Learning algorithm used for classification. It is based on Bayes’ Theorem and assumes that the features are independent of each other. The word “naive” means that the algorithm assumes each feature contributes independently to the final prediction. Gaussian Naive Bayes is used when the input features are continuous and follow a normal distribution. It uses statistical values such as mean, variance, and standard deviation to calculate the probability of a data point belonging to a particular class.

### 13. Bayes’ Theorem

Bayes’ Theorem is used to calculate the probability of an event based on prior knowledge and new evidence. It is an important concept in Naive Bayes classification because it helps calculate the probability that a data point belongs to a specific class.

Formula:

P(A|B) = [P(B|A) × P(A)] / P(B)

Where,
P(A|B) = Posterior probability
P(B|A) = Conditional probability or likelihood
P(A) = Prior probability
P(B) = Evidence

In classification, Bayes’ Theorem can be written as:

P(Class|x₁, x₂) = [P(x₁|Class) × P(x₂|Class) × P(Class)] / P(x₁, x₂)

Here, the model calculates the probability of each class for the given feature values and selects the class with the highest probability.

### 14. Prior Probability

Prior probability is the probability of a class before considering the new data. It is calculated from the training dataset. For example, if there are 20 total data records and 14 records belong to the Tech class, then the prior probability of Tech is:

P(Tech) = 14 / 20

Prior probability helps the model understand how common each class is before checking the feature values of the new data point.

### 15. Conditional Probability

Conditional probability is the probability of an event happening when another event is already known. In Gaussian Naive Bayes, conditional probability is used to calculate the probability of a feature value for a given class. For example, P(x₁|Tech) means the probability of feature x₁ occurring when the class is Tech. This helps the model decide how likely a new data point belongs to a particular class.

### 16. Joint Probability

Joint probability is the probability of two or more events happening together. In Naive Bayes classification, joint probability is calculated by multiplying the prior probability of the class with the conditional probabilities of the features.

Formula:

P(Class, x₁, x₂) = P(Class) × P(x₁|Class) × P(x₂|Class)

For example:

P(Tech, x₁, x₂) = P(Tech) × P(x₁|Tech) × P(x₂|Tech)

This gives the total probability score for the Tech class. The same process is repeated for other classes.

### 17. Posterior Probability

Posterior probability is the updated probability after considering the given evidence. In Naive Bayes classification, posterior probability is used to decide which class a new data point belongs to. The model calculates the posterior probability for each class and selects the class with the highest value.

Formula:

P(Class|x₁, x₂) = [P(x₁|Class) × P(x₂|Class) × P(Class)] / P(x₁, x₂)

If P(Tech|x₁, x₂) is greater than P(Business|x₁, x₂), then the data point is classified as Tech. If P(Business|x₁, x₂) is greater, then it is classified as Business.

### 18. Gaussian Probability Density Function

Gaussian Naive Bayes uses the Gaussian Probability Density Function to calculate the probability of continuous values. This formula is used when the data follows a normal distribution.

Formula:

P(x|μ, σ²) = 1 / √(2πσ²) × e^[-(x - μ)² / 2σ²]

Where,
x = Input value
μ = Mean
σ² = Variance
π = 3.1416
e = Euler’s number

For example, if x₁ = 78, mean = 90, and variance = 25, the Gaussian Probability Density Function can be used to calculate P(x₁ = 78|Tech). This probability shows how likely the value 78 belongs to the Tech class.

### 19. Mean, Median, Variance, and Standard Deviation

Mean, median, variance, and standard deviation are important statistical concepts used in Gaussian Naive Bayes. Mean is the average value of the data. Median is the middle value of the data when arranged in order. Variance shows how spread out the data values are from the mean. Standard deviation shows the average distance of data values from the mean. In Gaussian Naive Bayes, mean and variance are mainly used to calculate the probability of continuous feature values using the Gaussian formula.

### 20. Classification Example

In Gaussian Naive Bayes, a new data point is classified by calculating the probability score for each class. For example, if a new data point has values x₁ = 78 and x₂ = 42, the model calculates the probability score for each class such as Tech and Business. The score is calculated using the prior probability of the class and the conditional probability of each feature.

Formula:

Score = P(Class) × P(x₁|Class) × P(x₂|Class)

For Tech:

Score for Tech = P(Tech) × P(x₁|Tech) × P(x₂|Tech)

For Business:

Score for Business = P(Business) × P(x₁|Business) × P(x₂|Business)

After calculating both scores, the class with the highest score is selected as the final predicted class. If the score for Tech is higher, then the new data point belongs to Tech. If the score for Business is higher, then the new data point belongs to Business.

### 21. Confusion Matrix

A confusion matrix is used to evaluate the performance of a classification model. It shows how many predictions are correct and incorrect. It contains four main values: True Positive, True Negative, False Positive, and False Negative. True Positive means the model correctly predicted the positive class. True Negative means the model correctly predicted the negative class. False Positive means the model predicted positive, but the actual class was negative. False Negative means the model predicted negative, but the actual class was positive.

### 22. Evaluation Metrics

Evaluation metrics are used to measure how well a machine learning model performs. Common evaluation metrics include accuracy, precision, recall, and F1-score. These metrics are calculated from the confusion matrix.

Accuracy formula:

Accuracy = (TP + TN) / (TP + TN + FP + FN)

Precision formula:

Precision = TP / (TP + FP)

Recall formula:

Recall = TP / (TP + FN)

F1-Score formula:

F1-Score = 2 × (Precision × Recall) / (Precision + Recall)

Accuracy shows the overall correct predictions made by the model. Precision shows how many predicted positive results were actually correct. Recall shows how many actual positive results were correctly predicted. F1-score gives a balance between precision and recall.

```python
df_encoded['GENDER'] = df_encoded['GENDER'].map({'M':0,'F':1})
```

```python
df_encoded.head()
```

Encoding categorical data to numerical values 

```python
df_encoded['LUNG_CANCER'] = df_encoded['LUNG_CANCER'].map({'YES':1, 'NO':0})
```

Split data set into features(X) and target(Y), in our case, LUNG_CANCER column is the target and others are features

```python
X = df_encoded.drop('LUNG_CANCER', axis=1)
```

```python
y = df_encoded['LUNG_CANCER']
```

Split your data into train and test dataset. Import library 

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

Lets train the model 

```python
from sklearn.naive_bayes import GaussianNB
```

```python
model = GaussianNB()
```

Training the model

```python
model.fit(X_train, y_train)
```

Prediction on test data 

```python

y_pred = model.predict(X_test)
```

```python
Input_data = [
    0,  # GENDER: M (0) for Male
    55,  # AGE: 55
    1,  # SMOKING: 1 (Yes)
    2,  # YELLOW_FINGERS: 2 (High severity)
    2,  # ANXIETY: 2 (High anxiety)
    1,  # PEER_PRESSURE: 1 (Moderate pressure)
    1,  # CHRONIC DISEASE: 1 (Yes)
    2,  # FATIGUE: 2 (High fatigue)
    1,  # ALLERGY: 1 (Present allergy)
    2,  # WHEEZING: 2 (Severe wheezing)
    2,  # ALCOHOL CONSUMING: 2 (Alcohol consumption)
    2,  # COUGHING: 2 (Severe coughing)
    2,  # SHORTNESS OF BREATH: 2 (Severe shortness of breath)
    2,  # SWALLOWING DIFFICULTY: 2 (Severe difficulty)
    2   # CHEST PAIN: 2 (Severe chest pain)
]


prediction = model.predict([Input_data])  # Pass the list as 2D array (1 sample, 15 features)

# Print the prediction
if prediction == 1:
    print("The prediction is: LUNG CANCER (YES)")
else:
    print("The prediction is: NO LUNG CANCER")
```

Model Evaluation

```python
from sklearn.metrics import accuracy_score, confusion_matrix,classification_report
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

Model Evaluation

```python

accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.4f}")


print("Classification Report:")
print(classification_report(y_test, y_pred))

conf_matrix = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(conf_matrix)

# Visualizing the confusion matrix
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', xticklabels=['No Cancer', 'Cancer'], yticklabels=['No Cancer', 'Cancer'])
plt.title('Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```
